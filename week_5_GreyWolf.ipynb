```python
import numpy as np

# Traffic Signal Parameters
CYCLE_TIME = 120
MIN_GREEN = 10
MAX_GREEN = 110

ARRIVAL_RATE_1 = 0.35
ARRIVAL_RATE_2 = 0.25

SERVICE_HEADWAY = 2

# Generate fixed vehicle arrival patterns
np.random.seed(42)

TOTAL_TIME = CYCLE_TIME * 20

arrivals_1 = np.random.poisson(ARRIVAL_RATE_1, TOTAL_TIME)
arrivals_2 = np.random.poisson(ARRIVAL_RATE_2, TOTAL_TIME)


# Fitness Function
def fitness_function(green_time):

    green_1 = int(round(green_time))
    green_2 = CYCLE_TIME - green_1

    queue_1 = 0
    queue_2 = 0

    total_waiting_time = 0

    for t in range(TOTAL_TIME):

        cycle_position = t % CYCLE_TIME

        # Add incoming vehicles
        queue_1 += arrivals_1[t]
        queue_2 += arrivals_2[t]

        # Road 1 gets green signal
        if cycle_position < green_1:

            if cycle_position % SERVICE_HEADWAY == 0:
                if queue_1 > 0:
                    queue_1 -= 1

        # Road 2 gets green signal
        else:

            if (cycle_position - green_1) % SERVICE_HEADWAY == 0:
                if queue_2 > 0:
                    queue_2 -= 1

        # Calculate total waiting vehicles
        total_waiting_time += queue_1 + queue_2

    return total_waiting_time


# Grey Wolf Optimizer
def GWO(number_of_wolves, max_iterations):

    # Initialize wolf population
    wolves = np.random.uniform(
        MIN_GREEN,
        MAX_GREEN,
        number_of_wolves
    )

    alpha_position = 0
    beta_position = 0
    delta_position = 0

    alpha_score = float('inf')
    beta_score = float('inf')
    delta_score = float('inf')

    for iteration in range(max_iterations):

        # Evaluate fitness of each wolf
        for i in range(number_of_wolves):

            score = fitness_function(wolves[i])

            # Update Alpha
            if score < alpha_score:
                delta_score = beta_score
                delta_position = beta_position

                beta_score = alpha_score
                beta_position = alpha_position

                alpha_score = score
                alpha_position = wolves[i]

            # Update Beta
            elif score < beta_score:
                delta_score = beta_score
                delta_position = beta_position

                beta_score = score
                beta_position = wolves[i]

            # Update Delta
            elif score < delta_score:
                delta_score = score
                delta_position = wolves[i]

        # Decrease control parameter from 2 to 0
        a = 2 - iteration * (2 / max_iterations)

        # Update positions of wolves
        for i in range(number_of_wolves):

            r1 = np.random.random()
            r2 = np.random.random()

            A1 = 2 * a * r1 - a
            C1 = 2 * r2

            D_alpha = abs(C1 * alpha_position - wolves[i])
            X1 = alpha_position - A1 * D_alpha

            r1 = np.random.random()
            r2 = np.random.random()

            A2 = 2 * a * r1 - a
            C2 = 2 * r2

            D_beta = abs(C2 * beta_position - wolves[i])
            X2 = beta_position - A2 * D_beta

            r1 = np.random.random()
            r2 = np.random.random()

            A3 = 2 * a * r1 - a
            C3 = 2 * r2

            D_delta = abs(C3 * delta_position - wolves[i])
            X3 = delta_position - A3 * D_delta

            # Calculate new wolf position
            wolves[i] = (X1 + X2 + X3) / 3

            # Keep green time within limits
            wolves[i] = np.clip(
                wolves[i],
                MIN_GREEN,
                MAX_GREEN
            )

        print(
            f"Iteration {iteration + 1}: "
            f"Best Waiting Time = {alpha_score}"
        )

    return alpha_position, alpha_score


# Main Program
number_of_wolves = 20
max_iterations = 100

best_green_time, minimum_waiting_time = GWO(
    number_of_wolves,
    max_iterations
)

# Calculate green time for second road
best_green_time = round(best_green_time)
second_green_time = CYCLE_TIME - best_green_time

print("\n========== OPTIMIZATION RESULTS ==========")

print("Optimal Green Time for Road 1:",
      best_green_time, "seconds")

print("Optimal Green Time for Road 2:",
      second_green_time, "seconds")

print("Minimum Total Vehicle Waiting Time:",
      minimum_waiting_time, "vehicle-seconds")

print("Total Signal Cycle Time:",
      CYCLE_TIME, "seconds")


```
